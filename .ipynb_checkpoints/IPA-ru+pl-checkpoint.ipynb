```python
import pathlib
import os
from typing import List
from datasets import load_dataset, concatenate_datasets, ClassLabel
import evaluate
import torch
from transformers import (
    GPT2Tokenizer,
    Trainer,
    TrainingArguments,
    DataCollatorWithPadding,
)
from hf_wrapper import GPTForSequenceClassification
from model import GPT, GPTConfig
from tokenizer import load_tokenizer, eod_token
from sklearn.metrics import precision_score, recall_score, f1_score


def load_pretrained_model(path: pathlib.Path, device: str = 'cuda') -> GPT:
    checkpoint = torch.load(path, map_location=device)
    gptconf = GPTConfig(**checkpoint['model_args'])
    model = GPT(gptconf)
    state_dict = checkpoint['model']
    unwanted_prefix = '_orig_mod.'
    for k in list(state_dict.keys()):
        if k.startswith(unwanted_prefix):
            state_dict[k[len(unwanted_prefix):]] = state_dict.pop(k)
    filtered = {k: v for k, v in state_dict.items() if k in model.state_dict() and v.shape == model.state_dict()[k].shape}
    model.load_state_dict({**model.state_dict(), **filtered})
    return model.to(device)


# ---- Config ----
args = {
    'train_datasets': ['iggy12345/xnli-ru-ipa', 'iggy12345/cdsc-e-ipa'],
    'eval_datasets': {
        'ru': 'iggy12345/xnli-ru-ipa',
        'pl': 'iggy12345/cdsc-e-ipa'
    },
    'epochs': 3,
    'context_size': 1024,
    'learning_rate': 2e-5,
    'batch_size': 16,
    'hf_cache_dir': pathlib.Path('cache'),
    'device': 'cuda'
}

models_and_tokenizers = [
    {"model_type": "ipa",
     "model_path": "/fs/scratch/PAS2836/ipa_gpt/checkpoints/openwebtext_ipa_multi_node_12_5_50k/ckpt.pt",
     "tokenizer_paths": ("/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-ipa-number-preservation-vocab.json",
                         "/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-ipa-number-preservation-merges.txt")},
    {"model_type": "normal",
     "model_path": "/fs/scratch/PAS2836/ipa_gpt/checkpoints/openwebtext_normal_multi_node_12_5_50k/ckpt.pt",
     "tokenizer_paths": ("/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-normal-number-preservation-vocab.json",
                         "/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-normal-number-preservation-merges.txt")},
]

main_output_dir = pathlib.Path("./training_outputs_rupol")
os.makedirs(main_output_dir, exist_ok=True)

for config in models_and_tokenizers:
    model_type = config["model_type"]
    model_path = config["model_path"]
    vocab_path, merges_path = config["tokenizer_paths"]
    tokenizer = load_tokenizer(vocab_path, merges_path)

    output_dir = main_output_dir / f"output_{model_type}"
    os.makedirs(output_dir, exist_ok=True)

    print(f"\n🔄 Loading {model_type.upper()} model and tokenizer")
    base_model = load_pretrained_model(pathlib.Path(model_path), args['device'])
    base_model.config.pad_token_id = tokenizer.pad_token_id
    base_model.config.padding_side = tokenizer.padding_side
    model = GPTForSequenceClassification(base_model, num_classes=3).to(args['device'])

    def flatten_multi_features(examples, features: List[str]) -> List[str]:
        separator = f'\n\n{eod_token}\n\n'
        joined = []
        for items in zip(*[examples[f] for f in features]):
            safe_items = [(i if i is not None else "") for i in items]
            joined.append(separator.join(safe_items))
        return joined

    def preprocess_function(examples):
        if model_type == "ipa":
            if 'premise-phoneme' in examples:
                feature = flatten_multi_features(examples, ['premise-phoneme', 'hypothesis-phoneme'])
            else:
                feature = flatten_multi_features(examples, ['sentence_A-phoneme', 'sentence_B-phoneme'])
        else:
            if 'premise' in examples:
                feature = flatten_multi_features(examples, ['premise', 'hypothesis'])
            else:
                feature = flatten_multi_features(examples, ['sentence_A', 'sentence_B'])
        return tokenizer(feature, truncation=True, max_length=args['context_size'])

    # ---- Load and preprocess training datasets ----
    ru_dataset = load_dataset('iggy12345/xnli-ru-ipa', split='train', cache_dir=str(args['hf_cache_dir']))
    ru_label_classes = ru_dataset.features['label'].names

    pl_dataset = load_dataset('iggy12345/cdsc-e-ipa', split='train', cache_dir=str(args['hf_cache_dir']))
    pl_dataset = pl_dataset.cast_column('label', ClassLabel(names=ru_label_classes))

    full_train = concatenate_datasets([ru_dataset, pl_dataset])
    train_encoded = full_train.map(preprocess_function, batched=True)

    # ---- Load evaluation datasets ----
    eval_encoded = {}
    for lang_key, eval_ds in args['eval_datasets'].items():
        raw_eval = load_dataset(eval_ds, split="validation" if lang_key == "ru" else "train", cache_dir=str(args['hf_cache_dir']))
        eval_encoded[lang_key] = raw_eval.map(preprocess_function, batched=True)

    data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
    metric = evaluate.load("xnli", "en")

    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        predictions = torch.from_numpy(logits).argmax(dim=-1)
        return {
            **metric.compute(predictions=predictions, references=labels),
            "precision": precision_score(labels, predictions, average="macro"),
            "recall": recall_score(labels, predictions, average="macro"),
            "f1": f1_score(labels, predictions, average="macro")
        }

    training_args = TrainingArguments(
        output_dir=str(output_dir),
        eval_strategy="steps",
        eval_steps=500,
        save_strategy="steps",
        save_steps=500,
        save_total_limit=1,
        metric_for_best_model="f1",
        load_best_model_at_end=True,
        learning_rate=args['learning_rate'],
        per_device_train_batch_size=args['batch_size'],
        per_device_eval_batch_size=args['batch_size'],
        num_train_epochs=args['epochs'],
        weight_decay=0.01,
        logging_steps=500,
        logging_dir='./logs',
        fp16=True,
        disable_tqdm=False,
        warmup_ratio=0.3,
        save_safetensors=False
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=train_encoded,
        eval_dataset=eval_encoded['ru'],
        tokenizer=tokenizer,
        data_collator=data_collator,
        compute_metrics=compute_metrics,
    )

    print(f"\n🧠 Training {model_type.upper()} model on RU+PL")
    trainer.train(resume_from_checkpoint=False)

    for lang_key, dataset in eval_encoded.items():
        print(f"\n🔍 Evaluating on {lang_key.upper()} validation set:")
        results = trainer.evaluate(eval_dataset=dataset)
        print(f"✅ Results on {lang_key.upper()}:\n{results}")
```

```python
import os
import pathlib
import torch
from typing import List
from datasets import load_dataset, concatenate_datasets, ClassLabel
from transformers import GPT2Tokenizer, DataCollatorWithPadding, Trainer, TrainingArguments
from sklearn.metrics import precision_score, recall_score, f1_score
import evaluate

from tokenizer import load_tokenizer, eod_token
from model import GPT, GPTConfig
from hf_wrapper import GPTForSequenceClassification


def load_pretrained_model(path: pathlib.Path, device: str = 'cuda') -> GPT:
    checkpoint = torch.load(path, map_location=device)
    gptconf = GPTConfig(**checkpoint['model_args'])
    model = GPT(gptconf)
    state_dict = checkpoint['model']
    unwanted_prefix = '_orig_mod.'
    for k in list(state_dict.keys()):
        if k.startswith(unwanted_prefix):
            state_dict[k[len(unwanted_prefix):]] = state_dict.pop(k)
    filtered = {k: v for k, v in state_dict.items() if k in model.state_dict() and v.shape == model.state_dict()[k].shape}
    model.load_state_dict({**model.state_dict(), **filtered})
    return model.to(device)


args = {
    'hf_cache_dir': pathlib.Path('cache'),
    'context_size': 1024,
    'batch_size': 16,
    'device': 'cuda',
}

models_and_tokenizers = [
    {
        "model_type": "prebuilt",
        "model_path": "/fs/scratch/PAS2836/ipa_gpt/checkpoints/openwebtext_normal_multi_node_12_5_prebuilt/ckpt.pt",
        "tokenizer_name": "gpt2",
        "use_hf_tokenizer": True
    },
    {
        "model_type": "ipa",
        "model_path": "/fs/scratch/PAS2836/ipa_gpt/checkpoints/openwebtext_ipa_multi_node_12_5_50k/ckpt.pt",
        "tokenizer_paths": (
            "/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-ipa-number-preservation-vocab.json",
            "/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-ipa-number-preservation-merges.txt"
        ),
        "use_hf_tokenizer": False
    },
    {
        "model_type": "normal",
        "model_path": "/fs/scratch/PAS2836/ipa_gpt/checkpoints/openwebtext_normal_multi_node_12_5_50k/ckpt.pt",
        "tokenizer_paths": (
            "/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-normal-number-preservation-vocab.json",
            "/fs/ess/PAS2836/ipa_gpt/tokenizers/bpe-rus-pol-normal-number-preservation-merges.txt"
        ),
        "use_hf_tokenizer": False
    }
]


def flatten_multi_features(examples, features: List[str]) -> List[str]:
    separator = f'\n\n{eod_token}\n\n'
    joined = []
    for items in zip(*[examples[f] for f in features]):
        safe_items = [(i if i is not None else "") for i in items]
        joined.append(separator.join(safe_items))
    return joined


for config in models_and_tokenizers:
    model_type = config["model_type"]
    print(f"\n🔄 Loading {model_type.upper()} model and tokenizer")

    # Tokenizer
    if config.get("use_hf_tokenizer", False):
        tokenizer = GPT2Tokenizer.from_pretrained(config["tokenizer_name"])
        tokenizer.pad_token = tokenizer.eos_token
    else:
        vocab_path, merges_path = config["tokenizer_paths"]
        tokenizer = load_tokenizer(vocab_path, merges_path)

    # Model
    model_path = config["model_path"]
    base_model = load_pretrained_model(pathlib.Path(model_path), args['device'])
    base_model.config.pad_token_id = tokenizer.pad_token_id
    base_model.config.padding_side = tokenizer.padding_side
    model = GPTForSequenceClassification(base_model, num_classes=3).to(args['device'])

    def preprocess_fn(examples):
        if model_type == "ipa":
            if 'premise-phoneme' in examples:
                feature = flatten_multi_features(examples, ['premise-phoneme', 'hypothesis-phoneme'])
            else:
                feature = flatten_multi_features(examples, ['sentence_A-phoneme', 'sentence_B-phoneme'])
        else:
            if 'premise' in examples:
                feature = flatten_multi_features(examples, ['premise', 'hypothesis'])
            else:
                feature = flatten_multi_features(examples, ['sentence_A', 'sentence_B'])
        return tokenizer(feature, truncation=True, max_length=args['context_size'])

    # Load full eval datasets
    ru = load_dataset('iggy12345/xnli-ru-ipa', split='validation', cache_dir=str(args['hf_cache_dir']))
    label_classes = ru.features['label'].names  # ['entailment', 'neutral', 'contradiction']

# Load PL and cast the label column to match RU
    pl = load_dataset('iggy12345/cdsc-e-ipa', split='train', cache_dir=str(args['hf_cache_dir']))
    from datasets import ClassLabel
    pl = pl.cast_column('label', ClassLabel(names=label_classes))

# Now combine safely
    combined = concatenate_datasets([ru, pl])

    ru_encoded = ru.map(preprocess_fn, batched=True)
    pl_encoded = pl.map(preprocess_fn, batched=True)
    both_encoded = combined.map(preprocess_fn, batched=True)

    data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
    metric = evaluate.load("xnli", "en")

    def compute_metrics(eval_pred):
        logits, labels = eval_pred
        preds = torch.from_numpy(logits).argmax(dim=-1)
        return {
            **metric.compute(predictions=preds, references=labels),
            "precision": precision_score(labels, preds, average="macro"),
            "recall": recall_score(labels, preds, average="macro"),
            "f1": f1_score(labels, preds, average="macro"),
        }

    training_args = TrainingArguments(
        output_dir=f"./eval_only_{model_type}",
        per_device_eval_batch_size=args['batch_size'],
        do_train=False,
        do_eval=True,
        logging_dir="./logs",
        report_to=[],
    )

    trainer = Trainer(
        model=model,
        args=training_args,
        tokenizer=tokenizer,
        data_collator=data_collator,
        compute_metrics=compute_metrics
    )

    for lang, dataset in zip(['RU', 'PL', 'BOTH'], [ru_encoded, pl_encoded, both_encoded]):
        print(f"\n🔍 Evaluating {model_type.upper()} on {lang} dataset")
        results = trainer.evaluate(eval_dataset=dataset)
        print(f"✅ {model_type.upper()} on {lang}:")
        print(results)
```
